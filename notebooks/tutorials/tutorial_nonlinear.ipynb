# Nonlinear Poisson-Boltzmann

Se muestra un ejemplo del uso de la ecuacion de Poisson-Boltzmann no lineal para una proteina vista en el paper de Mauricio. La nolinealidad dentro del modelo de PB se ve representada por el siguiente termino

$$
N(\phi(\mathbf{x})) = \begin{cases} 
\sinh(\phi(\mathbf{x})) & \text{for nonlinear PBE} \\ 
\phi(\mathbf{x}) & \text{for linear PBE} 
\end{cases}
$$

En este tutorial se abarcara el calculo de potencial y la representacion de la superficie del potencial en un ARN cargado

```python
import os
os.chdir('../..')
```

```python
import pbj
```

Se carga el simulation en modo npbe y luego un soluto. El thickness corresponde a la zona del solvente donde se aplicara la nolinealidad, muy lejos del soluto se mantendra PB lineal

```python
simulation = pbj.implicit_solvent.Simulation(simulation_type='npbe')
simulation.add_solute('notebooks/tutorials/1ajf_nonlinear/1ajf.pqr',  
                      mesh_density=4,         
                      max_volume_size=0.1,
                      pb_formulation_thickness_width=3.0)
simulation.gmres_tolerance = 1e-6
```

Se utiliza el metodo `calculate_solvation_energy()` para el calculo de la energia usando el metodo de Newton-Raphson

```python
simulation.calculate_solvation_energy(units='kcal_mol')
```

Informacion relevante de la simulacion se puede recuperar usando los metodos `get_info()` y `get_results()`

```python
simulation.get_info()
```
